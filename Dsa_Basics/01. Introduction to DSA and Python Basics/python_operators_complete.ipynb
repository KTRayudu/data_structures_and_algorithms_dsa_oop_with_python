# Python Operators — Complete Reference

A single end-to-end notebook covering **all** Python operator categories, including a few that are commonly left out of summary lists:
- Bitwise **assignment** operators (`&=`, `|=`, `^=`, `<<=`, `>>=`)
- **Unary** operators (`+x`, `-x`, `~x`)
- The **walrus** operator (`:=`, Python 3.8+)

| Category | Operators |
|---|---|
| Arithmetic | `+  -  *  /  //  %  **` |
| Unary | `+x  -x  ~x` |
| Comparison | `==  !=  >  <  >=  <=` |
| Logical | `and  or  not` |
| Bitwise | `&  \|  ^  ~  <<  >>` |
| Assignment | `=  +=  -=  *=  /=  //=  %=  **=  &=  \|=  ^=  <<=  >>=` |
| Membership | `in  not in` |
| Identity | `is  is not` |
| Sequence | `+ (concat)  * (repeat)  [i]  [a:b]  in / not in` |
| Ternary | `value_if_true if condition else value_if_false` |
| Walrus (assignment expr) | `:=` |


## 1. Arithmetic Operators
Perform basic mathematical calculations.

```python
a, b = 17, 5

print("a + b  =", a + b)   # Addition
print("a - b  =", a - b)   # Subtraction
print("a * b  =", a * b)   # Multiplication
print("a / b  =", a / b)   # True division (float result)
print("a // b =", a // b)  # Floor division
print("a % b  =", a % b)   # Modulus (remainder)
print("a ** b =", a ** b)  # Exponentiation
```

## 2. Unary Operators
Operate on a single operand. Often left off operator lists but part of the arithmetic/bitwise family.

```python
x = 7

print("+x =", +x)   # Unary plus (no real effect on numbers)
print("-x =", -x)   # Unary minus (negation)
print("~x =", ~x)   # Bitwise NOT (inverts bits, equals -(x+1))
```

## 3. Comparison (Relational) Operators
Compare two values and return a boolean.

```python
a, b = 10, 20

print("a == b :", a == b)
print("a != b :", a != b)
print("a > b  :", a > b)
print("a < b  :", a < b)
print("a >= b :", a >= b)
print("a <= b :", a <= b)
```

## 4. Logical Operators
Combine conditional statements.

```python
x, y = True, False

print("x and y :", x and y)
print("x or y  :", x or y)
print("not x   :", not x)
```

## 5. Bitwise Operators
Operate on the binary representation of integers.

```python
a, b = 12, 10  # binary: 1100, 1010

print("a & b  =", a & b)   # AND
print("a | b  =", a | b)   # OR
print("a ^ b  =", a ^ b)   # XOR
print("~a     =", ~a)      # NOT
print("a << 2 =", a << 2)  # Left shift
print("a >> 2 =", a >> 2)  # Right shift
```

## 6. Assignment Operators
Assign and update variable values. Includes the commonly-omitted **bitwise compound assignment** operators.

```python
x = 10
print("x = 10        ->", x)

x += 5;  print("x += 5        ->", x)
x -= 3;  print("x -= 3        ->", x)
x *= 2;  print("x *= 2        ->", x)
x /= 4;  print("x /= 4        ->", x)
x //= 2; print("x //= 2       ->", x)
x **= 2; print("x **= 2       ->", x)
x %= 5;  print("x %= 5        ->", x)

y = 6
y &= 3;  print("y &= 3        ->", y)
y |= 8;  print("y |= 8        ->", y)
y ^= 2;  print("y ^= 2        ->", y)
y <<= 1; print("y <<= 1       ->", y)
y >>= 2; print("y >>= 2       ->", y)
```

## 7. Membership Operators
Test whether a value exists inside a sequence (list, string, tuple, set, dict).

```python
nums = [1, 2, 3, 4, 5]

print("3 in nums     :", 3 in nums)
print("9 not in nums :", 9 not in nums)
print("'py' in 'python' :", 'py' in 'python')
```

## 8. Identity Operators
Check whether two variables point to the exact same object in memory.

```python
a = [1, 2, 3]
b = a           # b references the same object as a
c = [1, 2, 3]   # c is a different object with equal content

print("a is b     :", a is b)
print("a is c     :", a is c)
print("a == c     :", a == c)      # equal value, different object
print("a is not c :", a is not c)
```

## 9. Sequence Operators
Operators that apply specifically to sequences (lists, tuples, strings): concatenation, repetition, indexing, slicing, and membership.

```python
s = "hello"
lst = [1, 2, 3]

print("s + ' world' =", s + " world")   # Concatenation
print("lst * 2      =", lst * 2)        # Repetition
print("s[1]         =", s[1])           # Indexing
print("s[1:4]       =", s[1:4])         # Slicing
print("'e' in s     =", 'e' in s)       # Membership (also listed above)
```

## 10. Ternary (Conditional) Operator
A compact one-line if-else expression.

```python
age = 20
status = "adult" if age >= 18 else "minor"
print("status =", status)
```

## 11. Walrus Operator `:=` (Python 3.8+)
Assignment **expression** — assigns a value to a variable as part of a larger expression. Frequently missing from operator lists but is a distinct operator category.

```python
data = [1, 2, 3, 4, 5, 6]

# Without walrus: compute length twice
if len(data) > 3:
    print("Long list, length =", len(data))

# With walrus: assign and use in the same expression
if (n := len(data)) > 3:
    print("Long list, length =", n)
```
